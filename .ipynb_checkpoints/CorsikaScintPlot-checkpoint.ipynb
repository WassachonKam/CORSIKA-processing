```python
import numpy as np
from functions import *
```

```python
def Ekin(px,py,pz,m):
    return np.sqrt(px**2 + py**2 + pz**2 + m**2) -m 

mumass = 0.105658 
emass = 5.11e-4

# kinnetic energy calculation all the units are in GeV
primary = ["proton"]
energy = ["lgE_16.0"]
sin2theta = ["0.0"]
runnum = 0


fNe = fp_Ne(primary[0], energy[0], sin2theta[0], runnum)
fNmu = fp_Nmu(primary[0], energy[0], sin2theta[0], runnum)
fileNe = np.load(fNe)
fileNmu = np.load(fNmu)
px_e, py_e, pz_e = fileNe['px'], fileNe['py'], fileNe['pz']
px_mu, py_mu, pz_mu = fileNmu['px'], fileNmu['py'], fileNmu['pz']

Ek_e = Ekin(px_e, py_e, pz_e, emass) 
Ek_mu = Ekin(px_mu, py_mu, pz_mu, mumass) 


with np.load(fNe) as data:
    all_arrays = dict(data.items())
all_arrays['Ekin'] = Ek_e
np.savez(fNe, **all_arrays)

with np.load(fNmu) as data:
    all_arrays = dict(data.items())
all_arrays['Ekin'] = Ek_mu
np.savez(fNmu, **all_arrays)
```

```python


def Eexpo(x):
    c = 0.3 # y offfset
    A = 1 # amplitute
    k = 3 #steepness
    x0 = -3 #threshold energy
    return c + A*( 1- np.exp(-k*(x-x0)))

def Elin(x):
    y0, y1 = 0.5e-3, 0.3
    x0, x1 = -4, -3
    return y0 * (y1/y0) ** ((x - x0)/(x1 - x0))
logEk = np.log10(Ek)

x = np.linspace(-4,1)
yarray = []
yEk = []

for i in x:
    # print(i)
    if i >= -3:
        y = Eexpo(i)
    else:
        y = Elin(i)
    yarray.append(y)

for i in logEk:
    if i >= -3:
        y = Eexpo(i)
    else:
        y = Elin(i)
    yEk.append(y)

plt.plot(x,yarray, label = 'empirical function')
plt.scatter(logEk, yEk, color = 'red', s  = 0.1, label = 'CORSIKA data')
plt.yscale('log')
plt.ylim(1e-4,1e2)
plt.xlim(-4,1)
plt.xlabel(rf'$\mathrm{{log_{{10}}(E_{{kin}}/GeV)}}$')
plt.ylabel(rf'$\mathrm{{E_{{deposited}}/MeV}}$')
plt.title("proton " + "lgE_16.0 " + "sin2_0.0 " + "run 0 " + rf"($e^{{\pm}}$)" )
plt.legend()
plt.grid()
```
